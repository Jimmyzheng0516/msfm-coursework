# Topic 05 - Time-Dependent Volatility, L2.29, and Practice Final Problem 2

This note teaches the L2.29 formula in detail and applies it to Practice Final Problem 2.

Main idea: if volatility is a known function of time only, then European option prices are still Black-Scholes prices, but with one effective volatility. That effective volatility is not the arithmetic average of volatility. It is the square root of average variance.

Course sources used: Lecture 2 slide L2.29, Homework 5 Problem 2, Practice Final Problem 2, and Practice Final Problem 2 solution.

## 1. What Problem Are We Solving?

In ordinary Black-Scholes, the stock has constant volatility:

$$dS_t=rS_tdt+\sigma S_tdW_t.$$

Here $\sigma$ is one fixed number, like $20\%$.

In this topic, volatility can change over calendar time:

$$dS_t=rS_tdt+\sigma(t)S_tdW_t.$$

This means:

- $S_t$ is the stock price at time $t$.
- $r$ is the risk-free interest rate.
- $W_t$ is Brownian motion under risk-neutral probabilities.
- $\sigma(t)$ is the instantaneous volatility at time $t$.

The key restriction is that $\sigma(t)$ is known in advance and is not random.

Example: volatility might be $20\%$ most of the time, but $80\%$ during a scheduled earnings announcement window.

That is exactly the idea in Practice Final Problem 2.

## 2. Instantaneous Volatility Versus Implied Volatility

These are not the same thing.

The model input $\sigma(t)$ is instantaneous volatility. It describes how noisy the stock is at each instant of calendar time.

The implied volatility $\sigma_{\mathrm{imp}}$ is the single constant volatility number that, when plugged into Black-Scholes, reproduces an option price.

So:

- $\sigma(t)$ is a time-by-time volatility schedule.
- $\sigma_{\mathrm{imp}}(K,T)$ is a quoted Black-Scholes volatility for a strike $K$ and expiry $T$.

In this special model, where volatility depends only on time and not on the stock price, all strikes with the same expiry have the same implied volatility.

So this model can create a term structure of implied volatility, meaning implied volatility changes with expiry $T$.

But this model cannot create a skew, meaning implied volatility changing with strike $K$ at the same expiry.

## 3. Why We Look At Log Stock Price

The stock equation is multiplicative:

$$dS_t=rS_tdt+\sigma(t)S_tdW_t.$$

The random shock is proportional to $S_t$. That makes $S_t$ itself less convenient to analyze directly.

So we look at:

$$\log S_t.$$

This is standard in Black-Scholes because geometric Brownian motion becomes normal after taking logs.

Using Itô's formula:

$$d\log S_t=\left(r-\frac{1}{2}\sigma(t)^2\right)dt+\sigma(t)dW_t.$$

The $-\frac{1}{2}\sigma(t)^2$ term is the Itô correction. It appears because $\log S_t$ is a curved function of $S_t$, not a linear function.

## 4. Where The Formula Comes From

Start from:

$$d\log S_t=\left(r-\frac{1}{2}\sigma(t)^2\right)dt+\sigma(t)dW_t.$$

Integrate both sides from time $0$ to time $T$:

$$\log S_T-\log S_0=\int_0^T \left(r-\frac{1}{2}\sigma(t)^2\right)dt+\int_0^T \sigma(t)dW_t.$$

Since $r$ is constant:

$$\int_0^T r\,dt=rT.$$

So:

$$\log S_T=\log S_0+rT-\frac{1}{2}\int_0^T\sigma(t)^2dt+\int_0^T\sigma(t)dW_t.$$

Now define the effective variance over $[0,T]$:

$$\bar{\sigma}_T^2T=\int_0^T\sigma(t)^2dt.$$

Equivalently:

$$\bar{\sigma}_T=\sqrt{\frac{1}{T}\int_0^T\sigma(t)^2dt}.$$

Then:

$$\log S_T=\log S_0+\left(r-\frac{1}{2}\bar{\sigma}_T^2\right)T+\int_0^T\sigma(t)dW_t.$$

## 5. Why The Random Integral Is Normal

The term

$$\int_0^T\sigma(t)dW_t$$

is a weighted sum of Brownian motion shocks.

If we approximate time by tiny intervals, this looks like:

$$\sigma(t_1)\Delta W_1+\sigma(t_2)\Delta W_2+\cdots+\sigma(t_n)\Delta W_n.$$

Each Brownian increment $\Delta W_i$ is normal. Different non-overlapping Brownian increments are independent.

A non-random weighted sum of independent normal random variables is normal.

Its mean is zero because each $\Delta W_i$ has mean zero.

Its variance is the sum of the variances:

$$\int_0^T\sigma(t)^2dt.$$

So:

$$\int_0^T\sigma(t)dW_t\sim N\left(0,\int_0^T\sigma(t)^2dt\right).$$

Using the definition of $\bar{\sigma}_T$:

$$\int_0^T\sigma(t)dW_t\sim N(0,\bar{\sigma}_T^2T).$$

## 6. The L2.29 Distribution Formula

Putting the pieces together:

$$\log S_T\sim N\left(\log S_0+\left(r-\frac{1}{2}\bar{\sigma}_T^2\right)T,\;\bar{\sigma}_T^2T\right),$$

where:

$$\bar{\sigma}_T=\sqrt{\frac{1}{T}\int_0^T\sigma(t)^2dt}.$$

This is the same lognormal structure as Black-Scholes with constant volatility $\bar{\sigma}_T$.

That is the main reason European option prices can still be computed by Black-Scholes.

## 7. The Plain English Meaning

$\bar{\sigma}_T$ is the single constant volatility that produces the same total variance from time $0$ to time $T$.

Total variance means:

$$\int_0^T\sigma(t)^2dt.$$

Average variance means:

$$\frac{1}{T}\int_0^T\sigma(t)^2dt.$$

Effective volatility means the square root of average variance:

$$\bar{\sigma}_T=\sqrt{\text{average variance}}.$$

This is why we say it is a root-mean-square volatility.

Important: you average $\sigma(t)^2$, not $\sigma(t)$.

## 8. Consequence For Black-Scholes Prices

A European call with strike $K$ and expiry $T$ has the same price as a Black-Scholes call with constant volatility $\bar{\sigma}_T$.

So:

$$C_0(K,T)=C^{BS}(S_0,0,K,T,r,r,\bar{\sigma}_T).$$

If the course writes Black-Scholes more compactly, this is just:

$$C_0(K,T)=C^{BS}(\bar{\sigma}_T).$$

Therefore the Black-Scholes implied volatility is:

$$\sigma_{\mathrm{imp}}(K,T)=\bar{\sigma}_T.$$

Notice there is no $K$ in the formula for $\bar{\sigma}_T$.

So, for a fixed expiry $T$, every strike gets the same implied volatility in this model.

## 9. Simple Example 1: Constant Volatility Check

Suppose $\sigma(t)=0.2$ for every $t$ from $0$ to $T$.

Then:

$$\int_0^T\sigma(t)^2dt=\int_0^T0.2^2dt=0.04T.$$

So:

$$\bar{\sigma}_T=\sqrt{\frac{1}{T}0.04T}=\sqrt{0.04}=0.2.$$

Good. If volatility is constant, the formula gives back the same constant volatility.

## 10. Simple Example 2: Two Volatility Blocks

Suppose:

$$\sigma(t)=0.3\quad\text{for }0\le t\le 0.5,$$

and:

$$\sigma(t)=0.1\quad\text{for }0.5<t\le 1.0.$$

Find the one-year effective volatility.

Step 1: square the volatilities.

$$0.3^2=0.09,\qquad 0.1^2=0.01.$$

Step 2: multiply each variance by the length of time it applies.

$$0.09\cdot 0.5+0.01\cdot 0.5=0.045+0.005=0.05.$$

Step 3: divide by total time $1.0$ and take square root.

$$\bar{\sigma}_1=\sqrt{0.05}\approx 0.2236.$$

Common wrong answer: $0.2$, the arithmetic average of $0.3$ and $0.1$.

That is wrong because option variance accumulates through $\sigma^2dt$, not $\sigma dt$.

## 11. Piecewise Constant Volatility Formula

Most exam questions use piecewise constant volatility.

If volatility equals $\sigma_1$ for length $\Delta t_1$, $\sigma_2$ for length $\Delta t_2$, and so on, then:

$$\bar{\sigma}_T=\sqrt{\frac{\sigma_1^2\Delta t_1+\sigma_2^2\Delta t_2+\cdots}{T}}.$$

The procedure is:

1. Draw the time line.
2. Mark the expiry window.
3. Count how much time is spent at each volatility.
4. Square the volatilities.
5. Multiply squared volatility by time length.
6. Add those total variances.
7. Divide by total time.
8. Take square root.

This procedure solves Practice Final Problem 2.

## 12. Forward Implied Volatility From Time $t$ To Expiry $T$

At time $t$, for an option expiring at time $T$, the remaining volatility is only the volatility from $t$ to $T$.

The effective remaining volatility is:

$$\bar{\sigma}_{t\to T}=\sqrt{\frac{1}{T-t}\int_t^T\sigma(s)^2ds}.$$

This is the time-$t$ implied volatility for expiry $T$ in this model.

Plain English: when time has already passed, do not include past volatility. Only average the remaining future variance.

## 13. Practice Final Problem 2: Set Up

The practice final says interest rates are zero and the stock has dynamics:

$$dS_t=\sigma(t)S_tdW_t,$$

with $S_0=100$.

The volatility schedule is:

$$\sigma(t)=0.8\quad\text{for }t\in[0.11,0.12]\cup[0.36,0.37],$$

and:

$$\sigma(t)=0.2\quad\text{for all other times}.$$

Interpretation:

- Most of the time volatility is $20\%$.
- During two short event windows, volatility jumps to $80\%$.
- Each event window has length $0.01$ years.

The five expiries are:

$$0.08,\quad 0.16,\quad 0.24,\quad 0.32,\quad 0.40.$$

We need to compare the effective volatilities for these expiries.

## 14. Practice Final Problem 2(a): Which Expiry Has Highest Time-0 Implied Volatility?

For each expiry $T$, compute:

$$\bar{\sigma}_T=\sqrt{\frac{1}{T}\int_0^T\sigma(t)^2dt}.$$

The variance levels are:

$$0.2^2=0.04,$$

$$0.8^2=0.64.$$

### Expiry $T=0.08$

This expires before the first high-volatility event.

So all $0.08$ years have volatility $0.2$.

$$\bar{\sigma}_{0.08}=0.2.$$

### Expiry $T=0.16$

This includes the first event $[0.11,0.12]$ of length $0.01$.

High-vol time: $0.01$.

Low-vol time: $0.16-0.01=0.15$.

$$\bar{\sigma}_{0.16}=\sqrt{\frac{0.8^2\cdot0.01+0.2^2\cdot0.15}{0.16}}.$$

$$=\sqrt{\frac{0.64\cdot0.01+0.04\cdot0.15}{0.16}}.$$

$$=\sqrt{\frac{0.0064+0.0060}{0.16}}=\sqrt{0.0775}\approx0.2784.$$

### Expiry $T=0.24$

This also includes only the first event.

High-vol time: $0.01$.

Low-vol time: $0.24-0.01=0.23$.

$$\bar{\sigma}_{0.24}=\sqrt{\frac{0.8^2\cdot0.01+0.2^2\cdot0.23}{0.24}}.$$

$$=\sqrt{\frac{0.0064+0.0092}{0.24}}=\sqrt{0.065}\approx0.2550.$$

### Expiry $T=0.32$

This includes only the first event. It expires before the second event starts at $0.36$.

High-vol time: $0.01$.

Low-vol time: $0.32-0.01=0.31$.

$$\bar{\sigma}_{0.32}=\sqrt{\frac{0.8^2\cdot0.01+0.2^2\cdot0.31}{0.32}}.$$

$$=\sqrt{\frac{0.0064+0.0124}{0.32}}=\sqrt{0.05875}\approx0.2424.$$

### Expiry $T=0.40$

This includes both events.

High-vol time: $0.02$.

Low-vol time: $0.40-0.02=0.38$.

$$\bar{\sigma}_{0.40}=\sqrt{\frac{0.8^2\cdot0.02+0.2^2\cdot0.38}{0.40}}.$$

$$=\sqrt{\frac{0.0128+0.0152}{0.40}}=\sqrt{0.07}\approx0.2646.$$

Comparison:

$$0.2000,\quad0.2784,\quad0.2550,\quad0.2424,\quad0.2646.$$

The highest is the $0.16$ expiry.

Answer to part (a): $T=0.16$.

## 15. Why $T=0.16$ Beats $T=0.40$

This is a good intuition check.

$T=0.16$ includes one high-volatility event in a short total period.

$T=0.40$ includes two high-volatility events, but it also includes much more low-volatility time.

The question is not how many events are included. The question is how concentrated the event variance is in the total expiry window.

For $T=0.16$:

$$\text{high-vol fraction}=\frac{0.01}{0.16}=6.25\%.$$

For $T=0.40$:

$$\text{high-vol fraction}=\frac{0.02}{0.40}=5\%.$$

The shorter option has a larger concentration of event volatility, so its average variance is higher.

## 16. Practice Final Problem 2(b): Time-0 Implied Volatility Of The 0.24-Expiry Call

From the calculation above, for $T=0.24$:

$$\bar{\sigma}_{0.24}=\sqrt{\frac{0.8^2\cdot0.01+0.2^2\cdot0.23}{0.24}}.$$

This is the answer format the professor accepts.

If a decimal is wanted:

$$\bar{\sigma}_{0.24}=\sqrt{0.065}\approx0.25495.$$

So the implied volatility is about $25.5\%$.

Answer to part (b):

$$\boxed{\sqrt{\frac{0.8^2\cdot0.01+0.2^2\cdot0.23}{0.24}}}.$$

## 17. Practice Final Problem 2(c): When Is The Time-$t$ Implied Vol Highest?

Now the option expiry is fixed at $T=0.24$.

We ask: among times $0\le t<0.24$, when is the time-$t$ implied volatility of this same $0.24$-expiry call highest?

At time $t$, past volatility no longer matters. Only remaining future volatility matters.

So the relevant formula is:

$$\bar{\sigma}_{t\to0.24}=\sqrt{\frac{1}{0.24-t}\int_t^{0.24}\sigma(s)^2ds}.$$

The expiry $0.24$ includes the first event $[0.11,0.12]$, but not the second event.

Think through time:

- Before $t=0.11$, the first event is still ahead.
- At $t=0.11$, the event is just about to begin, so the remaining window still includes the whole event, but less low-volatility time than before.
- After $t=0.11$, part of the high-volatility event has already passed, so the remaining average variance starts to fall.
- After $t=0.12$, the event is entirely in the past, so all remaining volatility is just $0.2$.

Therefore the maximum occurs at the start of the first event:

$$\boxed{t=0.11}.$$

This is the answer to part (c).

## 18. More Detailed Justification For Part (c)

This part says justification is not needed, but understanding it helps you remember the answer.

For $t<0.11$, the whole event $[0.11,0.12]$ is still ahead.

Remaining high-vol time is $0.01$.

Remaining low-vol time is:

$$(0.24-t)-0.01=0.23-t.$$

So average remaining variance is:

$$\frac{0.8^2\cdot0.01+0.2^2\cdot(0.23-t)}{0.24-t}.$$

As $t$ moves closer to $0.11$, we remove low-volatility time from the front, but the full high-volatility event remains. Removing low-volatility time raises the average variance.

For $0.11<t<0.12$, part of the high-volatility event has already passed. Now moving forward removes high-volatility time, which lowers the average variance.

So the peak is exactly at $t=0.11$.

## 19. Practice Final Problem 2(d): Implied Volatility At That Time

From part (c), the maximizing time is:

$$t=0.11.$$

The option expiry is still:

$$T=0.24.$$

So the remaining time is:

$$T-t=0.24-0.11=0.13.$$

From $0.11$ to $0.24$:

- High-volatility time is $0.01$, from $0.11$ to $0.12$.
- Low-volatility time is $0.12$, from $0.12$ to $0.24$.

Therefore:

$$\bar{\sigma}_{0.11\to0.24}=\sqrt{\frac{0.8^2\cdot0.01+0.2^2\cdot0.12}{0.13}}.$$

Simplify the variance:

$$0.8^2\cdot0.01=0.64\cdot0.01=0.0064,$$

$$0.2^2\cdot0.12=0.04\cdot0.12=0.0048.$$

Total remaining variance:

$$0.0064+0.0048=0.0112.$$

Average remaining variance:

$$\frac{0.0112}{0.13}\approx0.0861538.$$

Effective volatility:

$$\sqrt{0.0861538}\approx0.2935.$$

Answer to part (d):

$$\boxed{\sqrt{\frac{0.8^2\cdot0.01+0.2^2\cdot0.12}{0.13}}\approx29.35\%.}$$

## 20. Why The Time-$0.11$ Implied Vol Is Higher Than The Time-0 Implied Vol

For the $0.24$-expiry option, the time-0 implied volatility was:

$$\sqrt{\frac{0.8^2\cdot0.01+0.2^2\cdot0.23}{0.24}}\approx25.5\%.$$

At time $0.11$, the implied volatility is:

$$\sqrt{\frac{0.8^2\cdot0.01+0.2^2\cdot0.12}{0.13}}\approx29.35\%.$$

Why did it rise?

Because from time $0$ to time $0.11$, the option passed through low-volatility time. The high-volatility event is now closer and makes up a larger fraction of the remaining life of the option.

This is like saying: if an exam is tomorrow, the last day has a bigger effect on your remaining study intensity than it did a month ago.

## 21. General Exam Procedure For This Topic

When you see time-dependent deterministic volatility:

1. Write the effective volatility formula:

$$\bar{\sigma}_{a\to b}=\sqrt{\frac{1}{b-a}\int_a^b\sigma(t)^2dt}.$$

2. Draw the interval $[a,b]$.

3. Mark all high-volatility or low-volatility subintervals inside it.

4. Compute total variance:

$$\sum_i \sigma_i^2\Delta t_i.$$

5. Divide by total length $b-a$.

6. Take square root.

7. If asked which expiry has highest implied vol, compare average variances. You can compare before taking square roots.

8. If asked time-$t$ implied vol for fixed expiry, use only the remaining interval $[t,T]$.

## 22. Related HW5 Calibration Type

Homework 5 also uses L2.29 in the opposite direction.

Instead of being given $\sigma(t)$ and asked for implied volatility, you may be given implied volatilities and asked to build a piecewise constant $\sigma(t)$.

The identity is:

$$T\sigma_{\mathrm{imp}}(T)^2=\int_0^T\sigma(t)^2dt.$$

Suppose we know implied volatilities for maturities $T_1<T_2<T_3$.

If $\sigma(t)=\sigma_1$ on $[0,T_1]$, then:

$$\sigma_1^2=\sigma_{\mathrm{imp}}(T_1)^2.$$

If $\sigma(t)=\sigma_2$ on $(T_1,T_2]$, then:

$$T_2\sigma_{\mathrm{imp}}(T_2)^2=T_1\sigma_{\mathrm{imp}}(T_1)^2+(T_2-T_1)\sigma_2^2.$$

So:

$$\sigma_2=\sqrt{\frac{T_2\sigma_{\mathrm{imp}}(T_2)^2-T_1\sigma_{\mathrm{imp}}(T_1)^2}{T_2-T_1}}.$$

This is the same idea: total variance accumulates over time.

## 23. Common Mistakes

Mistake 1: Averaging volatility instead of variance.

Wrong:

$$\frac{1}{T}\int_0^T\sigma(t)dt.$$

Correct:

$$\sqrt{\frac{1}{T}\int_0^T\sigma(t)^2dt}.$$

Mistake 2: Forgetting the square root at the end.

The integral gives total variance. Implied volatility is the square root of average variance.

Mistake 3: Counting the wrong event length.

In Practice Final Problem 2, each high-volatility event has length $0.01$, not $0.1$.

Mistake 4: Including a future event that is after the option expiry.

The $0.24$ expiry does not include the event from $0.36$ to $0.37$.

Mistake 5: Including past volatility in a time-$t$ implied volatility.

At time $t$, use $[t,T]$, not $[0,T]$.

Mistake 6: Thinking more events always means higher implied volatility.

What matters is event variance divided by the total time window.

## 25. Exercise Answers

### Answer 1

High-vol period length: $0.25$.

Low-vol period length: $0.75$.

$$\bar{\sigma}_1=\sqrt{0.4^2\cdot0.25+0.2^2\cdot0.75}.$$

$$=\sqrt{0.16\cdot0.25+0.04\cdot0.75}.$$

$$=\sqrt{0.04+0.03}=\sqrt{0.07}\approx0.2646.$$

### Answer 2

The expiry $0.5$ includes the event $[0.3,0.35]$ of length $0.05$.

Low-vol time is $0.5-0.05=0.45$.

$$\bar{\sigma}_{0.5}=\sqrt{\frac{0.6^2\cdot0.05+0.2^2\cdot0.45}{0.5}}.$$

$$=\sqrt{\frac{0.36\cdot0.05+0.04\cdot0.45}{0.5}}.$$

$$=\sqrt{\frac{0.018+0.018}{0.5}}=\sqrt{0.072}\approx0.2683.$$

### Answer 3

At time $0.3$, the remaining interval is $[0.3,0.5]$.

High-vol time is $0.05$.

Low-vol time is $0.15$.

$$\bar{\sigma}_{0.3\to0.5}=\sqrt{\frac{0.6^2\cdot0.05+0.2^2\cdot0.15}{0.2}}.$$

$$=\sqrt{\frac{0.018+0.006}{0.2}}=\sqrt{0.12}\approx0.3464.$$

### Answer 4

Expiry $0.32$ includes only the first event.

High-vol time is $0.01$.

Low-vol time is $0.31$.

$$\bar{\sigma}_{0.32}=\sqrt{\frac{0.8^2\cdot0.01+0.2^2\cdot0.31}{0.32}}.$$

$$=\sqrt{\frac{0.0064+0.0124}{0.32}}=\sqrt{0.05875}\approx0.2424.$$

### Answer 5

At time $0.12$, the first event has just ended. For expiry $0.24$, all remaining volatility is $0.2$.

So:

$$\bar{\sigma}_{0.12\to0.24}=0.2.$$

You can also verify by formula:

$$\sqrt{\frac{0.2^2\cdot0.12}{0.12}}=0.2.$$

## 26. What You Should Memorize

Memorize the meaning, not just the symbols.

For deterministic time-dependent volatility:

$$\boxed{\bar{\sigma}_{a\to b}=\sqrt{\frac{1}{b-a}\int_a^b\sigma(t)^2dt}}.$$

Plain English:

Effective volatility equals square root of average variance over the relevant time interval.

Time-0 implied volatility for expiry $T$:

$$\boxed{\sigma_{\mathrm{imp}}(K,T)=\bar{\sigma}_{0\to T}}.$$

Time-$t$ implied volatility for expiry $T$:

$$\boxed{\sigma_{\mathrm{imp}}(t,T)=\bar{\sigma}_{t\to T}}.$$

Main exam rule:

Square vol, multiply by time length, add, divide by total time, square root.
