# Solution to Question 2, PS5

```python
import random

def polya_urn(n_draws, n_rep=30):
    results = []
    for _ in range(n_rep):
        red, green = 1, 1
        frac_4000 = frac_8000 = None

        for t in range(1, n_draws + 1):
            total = red + green
            if random.random() < red / total:
                red += 1
            else:
                green += 1

            # record fractions at 4000 and 8000
            if t == 4000:
                frac_4000 = red / (red + green)
            if t == 8000:
                frac_8000 = red / (red + green)

        results.append((frac_4000, frac_8000))

    return results

# Run the simulation and print results
frac = polya_urn(8000, n_rep=30)

print("Run | Fraction after 4000 | Fraction after 8000")
for i, (f4000, f8000) in enumerate(frac, 1):
    print(f"{i:2d}  | {f4000:.3f}            | {f8000:.3f}")
```
